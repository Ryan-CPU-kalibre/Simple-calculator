```python
number1 = int(input("Write your first number: \n"))

operation = input("Choose operation: (+, -, *, /) \n")
if not operation in ["+", "-", "*", "/"]:
  raise ValueError("Only these operations are allowed: (+, -, *, /)")

number2 = int(input("Write your second number: \n"))

def add(a, b):
  return a + b

def substract(a, b):
  return a - b

def multiply(a, b):
  return a * b

def divide(a, b):
  if b == 0:
    raise ZeroDivisionError("You can't divide by zero")
  else:
    return a / b

if operation == "+":
    print(add(number1, number2))

elif operation == "-":
    print(substract(number1, number2))

elif operation == "*":
    print(multiply(number1, number2))

elif operation == "/":
  print(divide(number1, number2))
```
